### **Define the Base Animal Class**

```
// Step 1: Define the Animal base class
public class Animal
{
    public string Species {get; set;}
    public string Name {get; set;}
    public string Sound {get; set;}
    public int Age {get; set;}

    public Animal (string species, string name, string sound, int age)
    {
        Species = species;
        Name = name;
        Sound = sound;
        Age = age;
    }

    public virtual void MakeNoise()
    {
        Console.WriteLine($"The {Species} says {Sound}");
    }
}

public class Dog : Animal
{
    public Dog (string name, int age) : base ("Dog", name, "Woof", age){}

    public void Bark(int times)
    {
        for (int i = 0; i < times; i++)
        Console.WriteLine($"{Species} says {Sound}");
    }

    public override void MakeNoise()
    {
        
    }

    public void Perform (int times)
    {
        Bark(times);
    }
}

public class Cat : Animal
{
    public Cat (string name, int age) : base ("Cat", name, "Meow", age){}

    public void Meow(int times)
    {
        for (int i = 0; i < times; i++)
        {
            Console.WriteLine($"{Species} says {Sound}");
        }
    }

    public override void MakeNoise()
    {
        
    }

    public void Perform(int times)
    {
        Meow(times);
    }
}

public class Bird : Animal
{
    public Bird (string name, int age) : base ("Bird", name, "Tweet", age){}

    public void Tweet(int times)
    {
        for (int i = 0; i < times; i++)
        Console.WriteLine($"{Species} says {Sound}");
    }

    public override void MakeNoise()
    {
        
    }

    public void Perform(int times)
    {
            Tweet(times);   
    }
}

Dog dog = new Dog ("Rusty", 2);
Cat cat = new Cat ("Leroy", 4);
Bird bird = new Bird ("Tweetiepie", 4);

public Animal[] animals = {dog, cat, bird};
foreach (Animal a in animals)
{
    dog.Perform(3); 
    cat.Perform(6);
    bird.Perform(2);
}
```
